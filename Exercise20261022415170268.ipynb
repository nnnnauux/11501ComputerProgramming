## 一、學習目標（Objectives）

透過本題，練習 Python 中 `*`、`**` 與 `return` 的多種用途，包括：

1. 使用 `*args` 收集不定數量的位置參數。

def add(*args):
    return sum(args)

print(add(1, 2, 3))       # 6
print(add(10, 20, 30, 40)) # 100

*args 會將傳入的位置參數收集成一個 tuple（元組）。
sum(args) 計算所有數字的總和。
return 將計算結果回傳給呼叫函數的地方。

2. 使用 `**kwargs` 收集不定數量的關鍵字參數。

def show_info(**kwargs):
    return kwargs

print(show_info(name="Amy", age=20))

輸出：

{'name': 'Amy', 'age': 20}

**kwargs 會將關鍵字參數收集成一個 dictionary（字典）。
name="Amy" 和 age=20 都會成為字典中的鍵值配對。

3. 使用 `*` 解包可迭代物件。

numbers = [1, 2, 3]
print(*numbers)

輸出：

1 2 3

也可以將清單解包後傳入函數：

def add(a, b, c):
    return a + b + c

numbers = [10, 20, 30]
print(add(*numbers))  # 60

*numbers 相當於將清單中的元素分別傳入 a、b、c。

4. 使用 `**` 解包字典。

def introduce(name, age):
    return f"我是{name}，今年{age}歲。"

person = {"name": "Amy", "age": 20}

print(introduce(**person))

輸出：

我是Amy，今年20歲。

**person 會將字典的鍵值配對解包成關鍵字參數，相當於：

introduce(name="Amy", age=20)

5. 理解 `*` 的乘法用途及 `**` 的次方用途。

print(3 * 4)    # 12
print(3 ** 4)   # 81
print("Hi" * 3) # HiHiHi

*：數值相乘，也能重複字串或序列。
**：計算次方，例如 3⁴=81。

6. 使用 `return` 回傳函數的計算結果。

def square(number):
    result = number ** 2
    return result

answer = square(5)
print(answer)  # 25

return 會將結果交還給呼叫函數的地方，並結束目前函數的執行。

如果只使用 print()，則是將內容顯示在畫面上，並不等於回傳結果。

## 二、題目說明（Problem Description）

請設計一個 Python 程式，建立學生分數分析系統。

程式必須能夠接收多筆分數，根據指定的倍率（multiplier）及加分值（bonus）調整分數，接著計算調整後的分數平方、總分及平均分數，最後回傳分析結果。

### Part 1：完成函數

請閱讀以下函數，理解各部分的功能。

```python
def analyze_scores(*scores, **options):
    """
    分析學生分數。

    Parameters:
        *scores: 接收不定數量的分數
        **options: 接收選擇性設定，例如 multiplier 和 bonus

    Returns:
        一個字典，包含調整後的分數、平方、總分及平均分數。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
```

### Part 2：呼叫函數

```python
scores = [60, 70, 80]

settings = {
    "multiplier": 1.1,
    "bonus": 5
}
```

def analyze_scores(*scores, **options):
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    scores_list = options.pop("scores_list", [])
    scores = scores + tuple(scores_list)

    adjusted = [s * multiplier + bonus for s in scores]
    squared = [s ** 2 for s in adjusted]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }


result = analyze_scores(
    multiplier=1.1,
    bonus=5,
    scores_list=[60, 70, 80]
)

print(result)

## 三、語法整理（Learning Focus）

| 語法 | 用途 |
|---|---|
| `*scores`（函數定義） | 將不定數量的位置參數收集成 tuple |
| `**options`（函數定義） | 將不定數量的關鍵字參數收集成 dictionary |
| `score * multiplier` | 乘法 |
| `score ** 2` | 次方 |
| `f(*scores)` | 將可迭代物件解包為位置參數 |
| `f(**settings)` | 將字典解包為關鍵字參數 |
| `return` | 將函數的結果回傳給呼叫端 |
