1. 0.1的二進位表示法為何? 使用markdown語法，列出計算過程和理由，並描述看到的現象。
計算過程在十進位中轉換小數為二進位，我們使用乘 2 取整法（乘以 2，取出整數部分，小數部分繼續乘 2）：
$0.1 \times 2 = 0.2$，整數部分為 $0$
$0.2 \times 2 = 0.4$，整數部分為 $0$
$0.4 \times 2 = 0.8$，整數部分為 $0$
$0.8 \times 2 = 1.6$，整數部分為 $1$（小數部分為 $0.6$）
$0.6 \times 2 = 1.2$，整數部分為 $1$（小數部分為 $0.2$）
接下來 $0.2 \rightarrow 0.4 \rightarrow 0.8 \rightarrow 1.6$ 開始循環。

理由與二進位結果理由：
就像十進位無法用有限位數精確表達 $\frac{1}{3}$（$0.3333...$）一樣，十進位的小數 $0.1$ 在二進位中是一個無限循環小數。
二進位表示：
$$0.1_{10} = 0.0001100110011001100110011001100110011001100110011001100..._2$$
科學記號形式為：$1.10011001100110011001100..._2 \times 2^{-4}$

看到的現象
因為電腦的記憶體空間有限，無法儲存無限位數，所以必須在某個地方「截斷」或「捨去」。這導致在 Python 中執行以下運算時，會出現微小的精確度誤差：
>>> 0.1 + 0.2
0.30000000000000004
這並不是 Python 的 Bug，而是所有遵循標準的程式語言在處理十進位浮點數時共同的面貌。

2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙
當遇到像 $0.1$ 這種在二進位中是無限循環小數的數字時，標準的硬體浮點數（基於 IEEE 754 標準）無法儲存無限多位，因此只能在有限的位元（雙精度的 53 位尾數）處進行截斷。在截斷和捨棄多餘位數時，IEEE 754 標準採用了「向最近偶數捨入」（Round to Nearest, Ties to Even）機制，也就是俗稱的「四捨六入五成雙」。

Python 如何解決「無限多位與精度誤差」的問題？如果我們需要高精度、甚至完全不會失真的計算，Python 提供了兩個內建模組來解決這個問題：

1. 使用 decimal 模組（自訂精確度與四捨六入五成雙）decimal 模組允許我們指定小數點後的位數，並且它預設就是採用「四捨六入五成雙」（ROUND_HALF_EVEN）的捨入規則。

from decimal import Decimal

# 使用字串初始化，避免一開始就受到二進位轉換的誤差
a = Decimal('0.1')
b = Decimal('0.2')

print(a + b)  # 輸出：0.3 (精確相等，不再是 0.30000000000000004)

2. 使用 fractions 模組（用分數精確表達）
如果不想受限於小數點位數，可以使用分數來表示。分數在電腦中是以「分子」與「分母」的整數儲存，因此完全沒有無限循環小數的誤差。

from fractions import Fraction

# 1/10 + 2/10
x = Fraction(1, 10)
y = Fraction(2, 10)

print(x + y)  # 輸出：3/10

總結
IEEE 754 的作法：遇到無限位時，靠「四捨六入五成雙」硬性截斷，這就是誤差的來源。
Python 的解法：若要避開這個問題，科學或金融計算上通常會改用 decimal 模組（控制精確位數與指定捨入法）或 fractions 模組（用分數完美保留）。

輸出結果與各參數意義sys.float_info 是一個包含了底層硬體與 Python 解譯器對浮點數（double）限制的結構化資料（通常對應 C 語言的 <float.h>）。常見的重要屬性說明如下：
max: 浮點數所能表示的最大正數（約 $1.79 \times 10^{308}$）。
min: 浮點數所能表示的最小正規正數（約 $2.22 \times 10^{-308}$）。
dig: 十進位的有效精確位數（通常為 15）。這代表雙精度浮點數至少能保證 15 位十進位數字的精確度。
epsilon (eps): 機器精度（Machine Epsilon）。即 $1.0$ 與大於 $1.0$ 的下一個可表示浮點數之間的差值（約 $2.22 \times 10^{-16}$）。小於此值的誤差在浮點數運算中會被忽略。
mant_dig: 尾數（Mantissa）所佔的二進位位數（53 位，包含隱含的最高位 1）。

5. 你的方法，對於一般兩個浮點數加法，在科學或金融上是好方法嗎？請思考。
金融計算（Financial / Accounting）
結論：使用 Decimal（十進位精確運算）是非常好的方法，而使用標準的 float 是極糟且危險的方法。
原因：在財務上，每一分錢都很重要，絕對不允許「累積誤差」（例如銀行利息計算、稅率拆帳）。如果使用二進位浮點數，經過幾萬筆交易後，誤差累積起來可能會造成帳目不平。Decimal 透過模擬十進位運算，雖然犧牲了計算速度與記憶體空間，但換取了絕對的精確性，這在金融法規與合規性上是必須的。

科學與工程計算（Scientific / Engineering）
結論：使用標準的 float（IEEE 754 雙精度）通常是最好的方法，而改用 Decimal 反而是不好的方法。
原因：科學計算（如物理模擬、大數據分析、機器學習矩陣運算）涉及海量的浮點數運算。標準 float 是直接由硬體層級（CPU 的 FPU）加速支援的，速度極快。物理世界的測量本身就帶有實驗誤差（Uncertainty），絕對精確的數字在物理上意義有限。科學家更關心的是「相對誤差」是否小於容許範圍，並透過數值分析方法（如控制條件數、使用穩定演算法）來控制誤差擴散。若在科學計算中全面改用軟體層級實作的 Decimal，效能可能會暴跌數十倍甚至數百倍，得不償失。
